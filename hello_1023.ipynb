#### P3001. n的阶乘
![](images/2026-11-08-17-20-11.png)
![](images/2026-11-08-17-20-18.png)

````python
# 递归函数：求 n 的阶乘
def fact(n):
  # 终点线：当数减到 1 时，直接交出 1，不再往下减
  if n <= 1:
    return 1
  # 核心公式：n 的阶乘 = n 乘以 (n - 1) 的阶乘
  return n * fact(n - 1)

# 读取输入并转成整数
n = int(input())

# 计算并打印结果
print(fact(n))
````

#### P3002. 斐波那契数列
![](images/2026-11-08-17-19-38.png)
![](images/2026-11-08-17-19-48.png)

````python
n = int(input())

# 如果求第 1 项或第 2 项，直接输出 1
if n <= 2:
  print(1)
else:
  # a 代表前前项，b 代表前一项，初始为 F(1)=1, F(2)=1
  a, b = 1, 1

  # 从第 3 项开始，一步步往后加，算到第 n 项
  for _ in range(3, n + 1):
    a, b = b, a + b  # 新的数字等于前两个相加，两个指针一起往后挪

  print(b)
````

#### P3003. 枚举子集
![](images/2026-11-08-17-18-34.png)
![](images/2026-11-08-17-18-50.png)

````python
n = int(input())  

path = []  # 记录当前已选择的元素（对应 LeetCode 里的 path）


def dfs(i):
  # 对应 LeetCode 78 的递归出口（所有数字决策完毕）
  if i > n:
    print(*path,end = ' \n')  # 到了叶子节点直接输出，空集正好输出空行
    return

  # 分支一：选当前数 nums[i]（对应 LeetCode 中的选分支，必须放前面）
  path.append(i)  # 1. 做出选择
  dfs(i + 1)  # 2. 深入下一层递归
  path.pop()  # 3. 回溯恢复现场

  # 分支二：不选当前数（对应 LeetCode 中的不选分支）
  dfs(i + 1)  


# 2. 从数字 1 开始递归
dfs(1)  
````

#### P3004.整数划分
![](images/2026-11-08-17-17-53.png)
![](images/2026-11-08-17-18-10.png)

````python
# 1. 一行读取两个正整数 n 和 k
n, k = map(int, input().split())

path = []  # 记录当前挑出来的数字（篮子）

# start: 当前数字最小能选多少（不能小于前一个数，防止重复）
# remain: 还剩下多少总和没凑够
def dfs(start, remain): 
  # 已经选够了 k 个数字
  if len(path) == k:
    # 如果此时正好把 n 全部用光（剩余总和为 0），就是一种合法划分
    if remain == 0:
      print(*path,end=' \n')  # 打印这 k 个数字（空格隔开）
    return

  # 如果选的数超过了剩余总和，就无法凑够了，提前结束
  # 枚举当前这一位可能填的所有数字 i
  for i in range(start, remain + 1):
    path.append(i)  # 1. 做出选择：把数字 i 装进篮子
    dfs(i, remain - i)  # 2. 走向下一步：下一个数从 i 开始起步，剩下要凑的总和减去 i[cite: 32]
    path.pop()  # 3. 恢复现场：把 i 拿出来，试下一个更大的数字


# 从最小正整数 1 开始选，初始需要凑的总和是 n[cite: 31]
dfs(1, n)
````
