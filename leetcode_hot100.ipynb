````python
#1. 两数之和
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        for i, x in enumerate(nums):  # x=nums[i]
            for j in range(i + 1, len(nums)):  # 枚举 i 右边的 j
                if x + nums[j] == target:  # 满足要求
                    return [i, j]  # 返回两个数的下标

if __name__ == '__main__':
    my_solution = Solution()

    test_nums = [2,7,11,15]
    test_target = 9
    result = my_solution.twoSum(test_nums,test_target)
    print('计算结果是：',result)
````

class Solution:
  def twoSum(self, nums: List[int], target: int) -> List[int]:
    idx = {}  # 第 1 步：创建一个空哈希表（字典）作为我们的“登记本”
    for j, x in enumerate(nums):  # 第 2 步：让大家依次排队走过，j 是位置编号，x 是手里的数字
      if ( target - x in idx):  # 第 3 步：灵魂拷问——我需要的搭档（target - x）已经在登记本里了吗？

      idx[x] = j  # 第 5 步：如果没找到，把我自己（数字 x 和下标 j）登记到哈希表里

````python
# 49. 字母异位词分组
class Solution:
  def groupAnagrams(self, strs: list[str]) -> list[list[str]]:
    d = {}  # 第 1 步：创建一个空的哈希表（字典），用来当我们的“分类储物柜”

    for s in strs:  # 第 2 步：遍历输入的每一个单词 s
      sorted_s = "".join(sorted(s))  # 第 3 步：制作“身份证”——把单词 s 的字母拆开排序，再拼回字符串
      if sorted_s not in d:  # 第 4 步：检查储物柜里有没有这个标签的抽屉
        d[sorted_s] = []  # 如果没有，就新开一个空的抽屉（列表）
      d[sorted_s].append(s)  # 第 5 步：把当前的原始单词 s 塞进对应的抽屉里
    return list(d.values())  # 第 6 步：收工！把储物柜里所有的抽屉内容打包吐出去
````

````python
#128. 最长连续序列
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        st = set(nums)  # 第1步：把 nums 转成哈希集合（自动去重，且查找极快）
        ans = 0         # 第2步：用来记录目前找到的最长连续序列的长度，初始为 0

        for x in st:    # 第3步：遍历哈希集合里的每一个数字 x
            
            # 第4步：核心灵魂——判断 x 是不是序列的“起点”
            # 如果 x 的“前一个数”(x - 1) 在集合里存在，说明 x 前面还有别的数字连着它，
            # 那 x 绝对不是起点！既然不是起点，我们就懒得理它，直接跳过（continue）。
            if x - 1 in st:  
                continue
            
            # 第5步：如果代码能走到这里，说明 x 是一个连续序列的【绝对起点】！
            y = x + 1          # 我们从 x 的下一个数（x+1）开始往后寻找
            while y in st:     # 只要 y 还在集合里，说明连续链条还没断
                y += 1         # y 继续往后看下一个数（x+2, x+3...）
            
            # 第6步：当 while 循环停下来时，说明 y 不在集合里了（链条断了）
            # 此时，从起点 x 到最后一个连续的数（y-1），总共有多少个数呢？
            # 答案就是：y - x 个数。
            ans = max(ans, y - x)  # 用 ans 记录下当前链条长度，并和之前的历史最大值比较，取较大者

        return ans  # 循环结束后，返回最长的长度
````

````python
#160. 相交链表

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        A,B = headA,headB   # 第 1 步：让指针 A 站在 A 链表的起跑线上，指针 B 站在 B 链表的起跑线上
        while A != B:   # 第 2 步：只要 A 和 B 没有在同一个节点相遇，就一直往前走
            A = A.next if A else headB  # 第 3 步：A 往前走一步；如果 A 走到头了（None），就立刻切换到 headB
            B = B.next if B else headA  # 第 4 步：B 往前走一步；如果 B 走到头了（None），就立刻切换到 headA
        return A    # 第 5 步：当循环结束时，返回 A（或者 B 都可以），它就是相交的起点！
````

````python
# 1. 这就是后台隐藏的 ListNode 定义（也就是你图片里看到的）
class ListNode:

  def __init__(self, val=0, next=None):
    self.val = val  # 当前房间的数值
    self.next = next  # 指向下一个房间的指针


# 2. 你的解题算法类
class Solution:

  def getIntersectionNode(
      self, headA: ListNode, headB: ListNode
  ) -> ListNode:
    A, B = headA, headB
    while A != B:
      A = A.next if A else headB
      B = B.next if B else headA
    return A


# ==========================================
# 3. 本地手动拼装链表（模拟 LeetCode 测试用例）
# ==========================================

# 第一步：先拼装它们共用的“公共尾巴”：8 -> 4 -> 5
c3 = ListNode(5)  # 最后一个节点 5
c2 = ListNode(4, c3)  # 节点 4，它的下一个是 c3
c1 = ListNode(8, c2)  # 节点 8，它的下一个是 c2（这就是相交的起点 c1！）

# 第二步：拼装链表 A 独占的前半部分：4 -> 1，然后把尾巴连到 c1 上
nodeA_1 = ListNode(1, c1)  # 值为 1 的节点，它的下一个直接指向交点 c1
headA = ListNode(4, nodeA_1)  # 链表 A 的头结点（值为 4）

# 第三步：拼装链表 B 独占的前半部分：5 -> 0 -> 1，然后把尾巴连到 c1 上
nodeB_2 = ListNode(1, c1)  # 值为 1 的节点，它的下一个指向交点 c1
nodeB_1 = ListNode(0, nodeB_2)  # 值为 0 的节点，它的下一个指向 nodeB_2
headB = ListNode(5, nodeB_1)  # 链表 B 的头结点（值为 5）


# ==========================================
# 4. 运行并验证结果
# ==========================================
sol = Solution()
result_node = sol.getIntersectionNode(headA, headB)

if result_node:
  print(f"成功找到相交节点！该节点的数值是: {result_node.val}")
else:
  print("没有找到相交节点。")
````

````python
#206. 反转链表
#方法一：迭代（双指针）    迭代：用循环（比如 while 或 for）重复做同一件事
class Solution:

  def reverseList(self, head: ListNode) -> ListNode:
    cur, pre = (
        head,
        None,
    )  # 第 1 步：cur 站在头结点上准备干活，pre 站在空（None）的地方（因为新链表的最后一个节点得指向 None）

    while cur:  # 第 2 步：只要 cur 还没走到头，就一直循环干活
      tmp = cur.next  # 第 3 步：【最关键的保护伞】用 tmp 暂存下一个节点，防止断网失联！
      cur.next = pre  # 第 4 步：【掉头反转】把当前节点的指针改向，指向身后的 pre
      pre = cur  # 第 5 步：pre 向前迈一步，站到刚才 cur 的位置上
      cur = tmp  # 第 6 步：cur 向前迈一步，走到刚才用 tmp 记下来的安全位置上

    return pre  # 第 7 步：当循环结束后，pre 刚好停在新链表的头结点上，把它返回！
````

````python
#234. 回文链表
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:     
    def middleNode(self,head):      #middleNode（快慢指针找中点)
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        return slow

    def reverseList(self,head):
        pre,cur = None,head
        while cur:
            temp = cur.next
            cur.next = pre
            pre = cur
            cur = temp
        return pre 

    def isPalindrome(self,head):
        mid = self.middleNode(head)
        head2 = self.reverseList(mid)
        while head2:
            if head.val != head2.val:
                return False
            head = head.next
            head2 = head2.next
        return True
    
````

````python
# 141. 环形链表
class Solution:

  def hasCycle(self, head: Optional[ListNode]) -> bool:
    # 1. 乌龟和兔子同时从起点（head）出发
    slow = fast = head

    # 2. 只要快指针的前方有路，就一直跑
    while fast and fast.next:
      slow = slow.next  # 乌龟每次走 1 步
      fast = fast.next.next  # 兔子每次走 2 步

      # 3. 关键判定：如果兔子和乌龟踩在了同一个房间里，说明被套圈了！
      if fast is slow:
        return True  # 证明有环，返回 True

    # 4. 如果快指针顺利跑到了终点（None），说明是直路，无环
    return False  
````

````python
# 142. 环形链表Ⅱ
#方法一：
class Solution(object):
    def detectCycle(self, head):
        fast = slow = head
        while True:
            if not (fast and fast.next):
                return
            fast, slow = fast.next.next, slow.next
            if fast is slow: 
                break
        fast = head
        while fast is not slow:
            fast, slow = fast.next, slow.next
        return fast
````

````python
#方法二
class Solution:

  def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
    # 1. 乌龟和兔子同时从起点 head 出发
    slow = fast = head

    # 2. 外层循环：判断前方是否有路可走
    while fast and fast.next:
      slow = slow.next  # 乌龟每次走 1 步
      fast = fast.next.next  # 兔子每次走 2 步

      # 3. 关键判定：兔子和乌龟在物理内存中踩在同一个节点上（第一次相遇）
      if fast is slow:

        # 4. 第二阶段：直接利用原有的 head 和 slow 一同起步向前走
        while slow is not head:  # 只要两者还没相遇（没走到入口）
          slow = slow.next  # 留在相遇点的 slow 每次走 1 步
          head = head.next  # 站在起点大门的 head 每次走 1 步

        # 5. 当两者相撞时，当前节点就是环入口，直接强行返回退出
        return slow

    # 6. 如果 while 正常跑完了没触发相遇，说明直路走到头，无环，返回 None
    return None
````

````python
#21. 合并两个有序链表
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next  
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        cur = dummy = ListNode()  # 用哨兵节点简化代码逻辑
        while list1 and list2:
            if list1.val < list2.val:
                cur.next = list1  # 1. 改变指针，把 list1 挂到新链表尾部
                list1 = list1.next  # 2. list1 指针后移，迎接下一个对比
            else:
                cur.next = list2  # 把 list2 挂到新链表尾部
                list2 = list2.next  # list2 指针后移
            cur = cur.next  # 3. 新链表的尾部指针后移
        cur.next = list1 or list2
        return dummy.next
````

````python
#2. 两数相加
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
        cur = dummy = ListNode()  # 哨兵节点
        carry = 0  # 进位值，初始没有进位，为 0
        while l1 or l2 or carry:
            s = carry  # 1. 当前位的起始和，先吃进上一位的进位值
            if l1:  # 如果 l1 还没走到尽头，才去读取数值
                s += l1.val
                l1 = l1.next  # 指针向后移一格

            if l2:  # 如果 l2 还没走到尽头，才去读取数值
                s += l2.val
                l2 = l2.next  # 指针向后移一格

            cur.next = ListNode(s % 10)  # 取余数：只存个位数
            carry = s // 10  # 整除取商：算出要向下一位传递的进位
            cur = cur.next  # 工人移位：游标走到最新创建的末尾节点
        
        return dummy.next
````

````python
#19. 删除链表的倒数第N个结点
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: ListNode | None, n: int) -> ListNode | None:
        left = right = dummy = ListNode(next=head)
        for _ in range(n):
            right = right.next  # 右指针先向右走 n 步
        while right.next:
            left = left.next
            right = right.next  # 左右指针一起走
        left.next = left.next.next  # 左指针的下一个节点就是倒数第 n 个节点
        return dummy.next
````

````python
#如果不用哨兵节点，代码具体该怎么写？
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: ListNode | None, n: int) -> ListNode | None:
        right = head
        for _ in range(n):
            right = right.next

        if not right:   # 如果 right 走了 n 步后直接变成了 None，说明链表总长恰好就是 n！
                        # 倒数第 n 个节点，正好就是开头的第 1 个节点（头节点）！
            return head.next    # 特殊处理：删除头节点，把第二个节点作为新头返回

        left = head
        while right.next:
            left = left.next
            right = right.next
        left.next = left.next.next
        return head
````

````python
#24. 两两交换链表中的节点
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        node0 = dummy = ListNode(next=head)  # 用哨兵节点简化代码逻辑
        node1 = head
        while node1 and node1.next:  # 至少有两个节点
            node2 = node1.next
            node3 = node2.next

            node0.next = node2  # 0 -> 2
            node2.next = node1  # 2 -> 1
            node1.next = node3  # 1 -> 3

            node0 = node1  # 下一轮交换，0 是 1
            node1 = node3  # 下一轮交换，1 是 3
        return dummy.next  # 返回新链表的头节点
````

````python
#138. 随机链表的复制
"""
# Definition for a Node.
class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random
"""

class Solution:
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':
        # 复制每个节点，把新节点直接插到原节点的后面
        cur = head
        while cur:
            cur.next = Node(cur.val, cur.next)
            cur = cur.next.next

        # 遍历交错链表中的原链表节点
        cur = head
        while cur:
            if cur.random:
                # 要复制的 random 是 cur.random 的下一个节点
                cur.next.random = cur.random.next
            cur = cur.next.next

        # 删除交错链表中的原链表节点，剩下的节点即为新链表
        cur = dummy = Node(0, head)
        while cur.next:
            # 删除原链表的节点，即当前节点的下一个节点
            # 如果要恢复原链表，见另一份代码【Python3 写法二】
            cur.next = cur.next.next
            cur = cur.next

        return dummy.next

````

````python
#148. 排序链表
class Solution:
    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = fast = head
        while fast and fast.next:
            pre = slow  # 记录 slow 的前一个节点
            slow = slow.next
            fast = fast.next.next
        pre.next = None  # 断开 slow 的前一个节点和 slow 的连接
        return slow

    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        cur = dummy = ListNode()  # 用哨兵节点简化代码逻辑
        while list1 and list2:
            if list1.val < list2.val:
                cur.next = list1  # 把 list1 加到新链表中
                list1 = list1.next
            else:
                cur.next = list2  # 把 list2 加到新链表中
                list2 = list2.next
            cur = cur.next
        cur.next = list1 if list1 else list2  # 拼接剩余链表
        return dummy.next

    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
         # 1. 递归出口：如果链表为空或者只有一个节点，无需排序
        if head is None or head.next is None:
            return head

        # 2. 剪断：找到中间节点 head2，并切断前后连接
        head2 = self.middleNode(head)

        # 3. 分治：递归排序前半截和后半截
        head = self.sortList(head)
        head2 = self.sortList(head2)

        # 4. 合并：把两段已经排好序的链表缝合起来
        return self.mergeTwoLists(head, head2)
````
